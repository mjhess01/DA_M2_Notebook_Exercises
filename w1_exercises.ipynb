```python
# Exercise Set 3

import pandas as pd

data = {
"Store": ["A", "A", "A", "B", "B", "C", "C", "C", "C"],
"Product": ["Apples", "Oranges", "Bananas", "Apples", "Bananas", "Oranges", "Apples", "Bananas", "Bananas"],
"Sales": [100, 80, 90, 120, 110, 60, 95, 85, 75],
"Profit": [30, 25, 20, 40, 35, 15, 32, 22, 18],
}

df = pd.DataFrame(data)
print(df.head())
```

```python
# Exercise 1

df.groupby('Store')[['Sales', 'Profit']].agg('mean')
```

```python
# Exercise 2
df.groupby('Product').agg({
    'Sales': 'max',
    'Profit': 'min'
})
```

```python
# Exercise 3

df.groupby('Store').agg(
    Avg_Sales=('Sales', 'mean'),
    Max_Profit=('Profit', 'max')
)
```

```python
# Exercise 4 (Bonus)

df.groupby(['Store', 'Product'])['Sales'].sum()
```

```python
# Exercise Set 4

import pandas as pd

df_customers = pd.DataFrame({
"CustomerName": ["Alice", "Bob", "Charlie"]
}, index=[101, 102, 103])

df_orders = pd.DataFrame({
"OrderTotal": [250, 300, 400]
}, index=[101, 102, 103])
```

```python
# Exercise 1

result = df_customers.join(df_orders)
print(result)
```

```python
# Exercise 2

df_cities = pd.DataFrame({
    'City': ['NYC', 'LA', 'Chicago']
}, index=[101, 102, 103])

result = df_customers.join(df_cities)
print(result)
```
